# Graphic approximation of the area under the  $f(x) = \frac{ 4 }{ 1 + x^2 }$ function curve using rectangles

```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.colors as mcolors
```

```python
NUMBER_COLORS = 10
colors_array = np.array(list(mcolors.CSS4_COLORS.keys())) # 148 colors
```

```python
def f(x):
	return 4 / (1 + x ** 2)
```

```python
def get_color(index, shift=10):
	return colors_array[(index % NUMBER_COLORS) % colors_array.size + shift]
```

```python
def plot(points, x_lower_limit = 0, x_upper_limit = 5):
	rectangle_width = (x_upper_limit - x_lower_limit) / points # number of rectangles = points

	# Generate x values
	x_values = np.linspace(x_lower_limit, x_upper_limit - rectangle_width, points)

	# Generate y values
	y_values = f(x_values)

	# Create the figure and axis
	fig, ax = plt.subplots()

	# Plot the curve
	ax.plot(x_values, y_values)

	rectangles = []
	for rectangle in enumerate(points):
		rectangle_height = y_values[rectangle]
		rectangle_left = rectangle * rectangle_width
		rectangle_bottom = 0
		rectangle_plot = plt.Rectangle((rectangle_left, rectangle_bottom), rectangle_width, rectangle_height, alpha=1, color=get_color(rectangle))
		rectangles.append(rectangle_plot)
		ax.add_patch(rectangle_plot)

	ax.set_xlim(x_lower_limit, x_upper_limit)
	ax.set_ylim(0, 5)

	ax.set_title('Rectangles under the curve')
	ax.set_xlabel('x')
	ax.set_ylabel('y')

	plt.show()
```

```python
number_rectangles = 1000
x_lower_limit = 0
x_upper_limit = 5
plot(number_rectangles, x_lower_limit, x_upper_limit)
```
