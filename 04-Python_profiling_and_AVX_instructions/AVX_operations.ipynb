# AVX (Advanced Vector Extensions) operations

## Scalar multiplication

$$
\vec{v} \times x = (v_0 \times x, v_1 \times x, \ldots, v_{n - 1} \times x)
$$

```python
import timeit
import numpy as np
```

```python
n = 1000
repetitions = 10
```

## Python arrays version (scalar multiplication)

```python
def run(n):
	x = 2
	v = [i for i in range(1, n)]

	for i in range(n - 1):
		v[i] = v[i] * x

	# print(v)

times = timeit.repeat(stmt="run(n)", number=repetitions, globals=globals())
average_time = sum(times) / repetitions
print(f"T_s: {average_time}")
```

### Python NumPy arrays version (scalar multiplication)

```python
def run(n):
	x = 2
	v = np.array(range(1, n))

	for i in range(n - 1):
		v[i] = v[i] * x

	# print(v)

times = timeit.repeat(stmt="run(n)", number=repetitions, globals=globals())
average_time = sum(times) / repetitions
print(f"T_numpy: {average_time}")
```

## Dot product

$$
\vec{ v_1 } \times \vec{ v_2 } = (v_{ 1,0 } \times v_{ 2,0 }) + (v_{ 1,1 } \times v_{ 2,1 }) +  \ldots + (v_{ 1,n - 1 } \times v_{ 2,n - 1 })
$$

### Python (dot product)

```python
import timeit
import numpy as np
```

```python
def run(n):
	vector_range = range(0, n)
	v1 = np.array(vector_range)
	v2 = np.array(vector_range)

	for i in vector_range:
		v2[i] = 1

	result = 0
	for i in vector_range:
		result += (v1[i] + 1) * v2[i]

	# print(result)

times = timeit.repeat(stmt="run(n)", number=repetitions, globals=globals())
average_time = sum(times) / repetitions
print(f"Average time: {average_time}")
```
