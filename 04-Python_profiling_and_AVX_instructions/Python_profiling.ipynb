# Python profiling

## `timeit`

### Examples (`timeit`)

#### First example (`timeit`)

```python
import timeit


def run():
	ouput = ""

	for i in range(1000):
		ouput += str(i)

time = timeit.timeit(stmt=run, number=100000)

print(time)
```

```python
import timeit


starting_time = timeit.default_timer()
ouput = ""
for i in range(1000):
	ouput += str(i)
ending_time = timeit.default_timer()
print(f"Time elapsed: {ending_time - starting_time}")
```

> It is preferable to use `timeit.default_timer()` over `time.time()`, because `time.time()` relies on a system clock that can be adjusted by external factors (it is non-monotonic).

#### Second example (`timeit`)

```python
import timeit
from numpy import sqrt


starting_time = timeit.default_timer()
my_list = [sqrt(i) for i in range(50000)]
ending_time = timeit.default_timer()
print(f"Time elapsed: {ending_time - starting_time}")
```

## `cProfile`

```python
import cProfile as profile
from pstats import Stats, SortKey


prof = profile.Profile()


prof.enable()
ouput = ""
for x in range(10000):
	ouput += str(x)
prof.disable()


stats = Stats(prof).strip_dirs().sort_stats(SortKey.CUMULATIVE)
stats.print_stats(10)
```
